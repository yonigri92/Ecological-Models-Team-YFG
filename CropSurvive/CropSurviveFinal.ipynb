```python
#prediction methods
def predict_Trendline_one_value_for_key(current_hash_data,time_to_predict,skip_value):
#{
    new_hash_Data = current_hash_data.copy()
    all_bacteria_types = set([key[1] for key in new_hash_Data.keys()])

    for bacteria_type in all_bacteria_types:
      historical_days = []
      historical_values = []
      for (day, b_type), obj in current_hash_data.items():
        if b_type == bacteria_type:
            values_list = list(vars(obj).values())[2]
            valid_values = [v for v in values_list if pd.notna(v)]
            if valid_values:
                    historical_days.append(day)
                    historical_values.append(sum(valid_values) / len(valid_values))
        #now we have all the days and their memoza
        n = len(set(historical_days))
        if n == 0:
            continue #no data


        if n >= 2:
            #only 2 days = linear more that 2 days exp = 2
            degree = 2 if n >= 3 else 1
            coefficients = np.polyfit(historical_days, historical_values, degree)
        for i in range(1, time_to_predict + 1):
            daytime = i * skip_value
            if daytime <= MAX_REAL_DAY:
                continue
            if n >= 2:
                predicted_value = np.polyval(coefficients, daytime)
            else:
                predicted_value = historical_values[0]

            new_hash_Data[(daytime, bacteria_type)] = predicted_value

    return new_hash_Data
#}
#def predict_Trendline_one_value_for_key(current_hash_data,time_to_predict,skip_value):
def predict_Trendline_two_values_per_key(current_hash_data,time_to_predict,skip_value):
    #{ same thing as befor but not we do it for two values per key
    new_hash_Data = current_hash_data.copy()
    all_bacteria_types = set([key[1] for key in new_hash_Data.keys()])

    for bacteria_type in all_bacteria_types:
        historical_days_1 = []
        historical_values_1 = []
        historical_days_2 = []
        historical_values_2 = []

        for (day, b_type), obj in current_hash_data.items():
            if b_type == bacteria_type:
                attributes = list(vars(obj).values())

                values_list_1 = attributes[2]
                values_list_2 = attributes[3]

                valid_val_1 = [v for v in values_list_1 if pd.notna(v)]
                if valid_val_1:
                    historical_days_1.append(day)
                    historical_values_1.append(sum(valid_val_1) / len(valid_val_1))

                valid_val_2 = [v for v in values_list_2 if pd.notna(v)]
                if valid_val_2:
                    historical_days_2.append(day)
                    historical_values_2.append(sum(valid_val_2) / len(valid_val_2))

        n1 = len(set(historical_days_1))
        n2 = len(set(historical_days_2))

        if n1 >= 2:
            degree_1 = 2 if n1 >= 3 else 1
            coeff_1 = np.polyfit(historical_days_1, historical_values_1, degree_1)

        if n2 >= 2:
            degree_2 = 2 if n2 >= 3 else 1
            coeff_2 = np.polyfit(historical_days_2, historical_values_2, degree_2)

        for i in range(1, time_to_predict + 1):
            daytime = i * skip_value
            if daytime <= MAX_REAL_DAY:
                continue
            if n1 >= 2:
                pred_1 = np.polyval(coeff_1, daytime)
            elif n1 == 1:
                pred_1 = historical_values_1[0]
            else:
                pred_1 = 0

            if n2 >= 2:
                pred_2 = np.polyval(coeff_2, daytime)
            elif n2 == 1:
                pred_2 = historical_values_2[0]
            else:
                pred_2 = 0
            new_hash_Data[(daytime, bacteria_type)] = (pred_1, pred_2)

    return new_hash_Data
    #}
#def predict_Trendline_two_values_per_key(current_hash_data,time_to_predict,skip_value):
```

```python


#get XL data from database

dict_weight = collect_data_from_xl_files_weight()
dict_texture = collect_data_from_xl_files_texture_analyzer()
dict_tss = collect_data_from_xl_files_tss()
dict_o2 = collect_data_from_xl_files_o2()
dict_co2 = collect_data_from_xl_files_co2()
dict_morphology = collect_data_from_xl_files_morphology()
dict_bac_yeast = collect_data_from_xl_files_bacteria_total_yeasts()
dict_bac_lactic = collect_data_From_Xl_files_bacteria_total_lactic()
dict_color = collect_data_from_xl_files_color()
class UnifiedRecord:
    def __init__(self):
       self.weight_loss = None
       self.co2 = None
       self.o2 = None
       self.TSS = None
       self.texture_firmness = None
       self.color = None
       self.bacteria_total = None
       self.yeasts = None
       self.lactic = None
#the idea here is that instead of haveing 5 different variables we combine all of the sets into one big set
#its possible as all sets share keys (date(day),type of bacteria used)
#after that method we only need to use this single variable for all data
#it will also allow us to reset the data after we predict the future incase we want too
def build_unified_data(d_weight, d_texture, d_tss, d_o2, d_co2, d_morph, d_bac_y, d_bac_l, d_color):
    unified_data = {}

    all_keys = (set(d_weight.keys()) | set(d_texture.keys()) | set(d_tss.keys()) |
                set(d_o2.keys()) | set(d_co2.keys()) | set(d_morph.keys()) |
                set(d_bac_y.keys()) | set(d_bac_l.keys()) | set(d_color.keys()))

    for key in all_keys:
        rec = UnifiedRecord()


        if key in d_weight:
            val = d_weight[key]
            rec.weight_loss = val.weight_loss if hasattr(val, 'weight_loss') else val
        if key in d_co2:
            val = d_co2[key]
            rec.co2 = val.co2 if hasattr(val, 'co2') else val
        if key in d_o2:
            val = d_o2[key]
            rec.o2 = val.o2 if hasattr(val, 'o2') else val
        if key in d_tss:
            val = d_tss[key]
            rec.TSS = val.TSS if hasattr(val, 'TSS') else val
        if key in d_texture:
            val = d_texture[key]
            rec.texture_firmness = val.force if hasattr(val, 'force') else val
        if key in d_morph:
            val = d_morph[key]
            rec.morphology_rate = val.morphology_rate if hasattr(val, 'morphology_rate') else val
        if key in d_color:
            val = d_color[key]
            rec.color = val.color if hasattr(val, 'color') else val


        if key in d_bac_y:
            val = d_bac_y[key]
            if hasattr(val, 'bacteria_total'):
                rec.bacteria_total = val.bacteria_total
                rec.yeasts = val.yeasts
            else:
                rec.bacteria_total = val[0]
                rec.yeasts = val[1]

        if key in d_bac_l:
            val = d_bac_l[key]
            if hasattr(val, 'lactic'):
                rec.bacteria_total = val.bacteria_total
                rec.lactic = val.lactic
            else:
                rec.bacteria_total = val[0]
                rec.lactic = val[1]

        unified_data[key] = rec

    return unified_data

current_hash_data = build_unified_data(dict_weight, dict_texture, dict_tss, dict_o2, dict_co2, dict_morphology, dict_bac_yeast, dict_bac_lactic, dict_color)
```

```python


def get_avg_val(val):
    if val is None: return None
    if isinstance(val, list):
        valid = [v for v in val if pd.notna(v)]
        return sum(valid) / len(valid) if valid else None
    return val
#def get_avg_val(val):
def calculate_freshness_score(day, treatment, obj, full_data):

    NUM_METRICS = 7  # (Weight loss, CO2, O2, Texture, Color, Morphology, Bacteria total)


    MAX_PENALTY = 100.0 / NUM_METRICS

    day0_obj = full_data.get((0, treatment))
    co2_day0 = get_avg_val(getattr(day0_obj, 'co2', None)) if day0_obj else None
    o2_day0 = get_avg_val(getattr(day0_obj, 'o2', None)) if day0_obj else None
    color_day0 = get_avg_val(getattr(day0_obj, 'color', None)) if day0_obj else None
    morphology_day0 = get_avg_val(getattr(day0_obj, 'morphology_rate', None)) if day0_obj else None
    bacteria_day0 = get_avg_val(getattr(day0_obj, 'bacteria_total', None)) if day0_obj else None

    weight_loss = get_avg_val(getattr(obj, 'weight_loss', None))
    co2 = get_avg_val(getattr(obj, 'co2', None))
    o2 = get_avg_val(getattr(obj, 'o2', None))
    texture = get_avg_val(getattr(obj, 'texture_firmness', None))
    color = get_avg_val(getattr(obj, 'color', None))
    morphology = get_avg_val(getattr(obj, 'morphology_rate', None))
    bacteria = get_avg_val(getattr(obj, 'bacteria_total', None))

    if all(v is None for v in [weight_loss, co2, o2, texture, color, morphology, bacteria]):
        return None

    score = 100.0


    if weight_loss is not None and weight_loss > 0:
        weight_deg_pct = weight_loss * 10
        score -= min(MAX_PENALTY, weight_deg_pct * (MAX_PENALTY / 100))


    if co2 is not None and co2_day0 and co2_day0 > 0:
        co2_change_pct = ((co2 - co2_day0) / co2_day0) * 100
        if co2_change_pct > 0:
            score -= min(MAX_PENALTY, co2_change_pct * (MAX_PENALTY / 100))


    if o2 is not None and o2_day0 and o2_day0 > 0:
        o2_loss_pct = ((o2_day0 - o2) / o2_day0) * 100
        if o2_loss_pct > 0:
            score -= min(MAX_PENALTY, o2_loss_pct * (MAX_PENALTY / 100))


    if texture is not None and day0_obj:
        texture_day0 = get_avg_val(getattr(day0_obj, 'texture_firmness', None))
        if texture_day0 and texture_day0 > 0:
            texture_loss_pct = ((texture_day0 - texture) / texture_day0) * 100
            if texture_loss_pct > 0:
                score -= min(MAX_PENALTY, texture_loss_pct * (MAX_PENALTY / 100))


    if color is not None and color_day0 and color_day0 > 0:
        color_change_pct = ((color - color_day0) / color_day0) * 100
        if color_change_pct > 0:
            score -= min(MAX_PENALTY, color_change_pct * (MAX_PENALTY / 100))


    if morphology is not None and morphology_day0 and morphology_day0 > 0:
        morph_change_pct = ((morphology - morphology_day0) / morphology_day0) * 100
        if morph_change_pct > 0:
            score -= min(MAX_PENALTY, morph_change_pct * (MAX_PENALTY / 100))


    if bacteria is not None and bacteria_day0 and bacteria_day0 > 0:
        bacteria_change_pct = ((bacteria - bacteria_day0) / bacteria_day0) * 100
        if bacteria_change_pct > 0:
            score -= min(MAX_PENALTY, bacteria_change_pct * (MAX_PENALTY / 100))

    return max(0.0, min(100.0, score))
#def calculate_freshness_score(obj):
def update_freshness_graph():
    data_list = []
    for (day, treatment), obj in current_hash_data.items():
        fresh_score= calculate_freshness_score(day, treatment, obj, current_hash_data)
        if fresh_score is not None:
            data_list.append({"day": day, "treatment": treatment, "value": fresh_score})

    df = pd.DataFrame(data_list)
    if df.empty: return px.line(title="No Data")


    df = df.sort_values(by=["treatment", "day"])

    df['value'] = df.groupby('treatment')['value'].cummin()
    control_scores = dict(zip(df[df['treatment'].str.lower() == 'control']['day'], df[df['treatment'].str.lower() == 'control']['value']))
    df['value'] = df.apply(lambda r: r['value'] - control_scores.get(r['day'], r['value']) if r['day'] in control_scores else 0.0, axis=1)

    df = df.sort_values(by="day")
    df['Data Type'] = df['day'].apply(lambda d: 'Real Data' if d <= MAX_REAL_DAY else 'Prediction')
    fig = px.line(df, x="day", y="value", color="treatment", line_dash="Data Type", markers=True,
                  title="Freshness to Time", line_dash_map={'Real Data': 'solid', 'Prediction': 'dash'})
    fig.update_layout(height=600, xaxis_title="Time (Days)", yaxis_title="Freshness Difference from Control")
    return fig
#def update_freshness_graph():
def generate_freshness_heatmap():
    data_list = []
    for (day, treatment), obj in current_hash_data.items():
        score = calculate_freshness_score(day, treatment, obj, current_hash_data)
        if score is not None:
            data_list.append({"day": int(day), "treatment": str(treatment), "score": score})

    df = pd.DataFrame(data_list)
    if df.empty: return px.imshow([[0]], title="No Data")
    #start showing graph only from the first day where we have all the data on
    valid_days_count = df.groupby('day')['treatment'].nunique()
    max_treatments = df['treatment'].nunique()
    full_days = valid_days_count[valid_days_count == max_treatments].index
    if not full_days.empty:
        first_valid_day = full_days.min()
        df = df[df['day'] >= first_valid_day]
    ######
    df = df.sort_values(by=["treatment", "day"])
    df['score'] = df.groupby('treatment')['score'].cummin()
    control_scores_hm = dict(zip(df[df['treatment'].str.lower() == 'control']['day'], df[df['treatment'].str.lower() == 'control']['score']))
    df['score'] = df.apply(lambda r: r['score'] - control_scores_hm.get(r['day'], r['score']) if r['day'] in control_scores_hm else 0.0, axis=1)

    pivot_df = df.pivot(index="treatment", columns="day", values="score")
    row_order = ["control", "CMC-3-methyl", "CMC-3-methyl+A", "CMC-3-methyl+B", "CMC-3-methyl+C"]
    existing_order = [t for t in row_order if t in pivot_df.index] + [t for t in pivot_df.index if t not in row_order]
    pivot_df = pivot_df.reindex(existing_order)
    fig = px.imshow(pivot_df,
                    labels=dict(x="Time (Days)", y="Treatment type", color="Freshness score"),
                    x=pivot_df.columns,
                    y=pivot_df.index,
                    color_continuous_scale="RdYlGn",

                    text_auto=".1f",
                    aspect="auto",
                    title="")

    fig.update_layout(height=600, xaxis=dict(tickmode='linear'))
    return fig
```

```python
def calculate_empirical_pca_pressure():
    scores = []
    for (day, treatment), obj in current_hash_data.items():
        if day == 7:
            s = calculate_freshness_score(day, treatment, obj, current_hash_data)
            if s is not None: scores.append(s)

    if not scores: return 1.0
    avg_freshness = sum(scores) / len(scores)
    pressure = 2.0 - (avg_freshness / 100.0)
    return round(pressure, 2)
```

```python
class MicrobialCompetitionABM:
    def __init__(self, width=40, height=40, yeast_count=20, lactic_count=20, pca_pressure=1.0):
        self.width = width
        self.height = height
        self.grid = np.zeros((height, width), dtype=int)
        self.health_grid = np.full((height, width), 100.0)
        self.acid_grid = np.zeros((height, width))
        self.entities = {}
        self.pca_pressure = pca_pressure

        for _ in range(int(yeast_count)):
            self._spawn('yeast', 1, init_energy=10)


        for _ in range(int(lactic_count)):
            self._spawn('lactic', 2, init_energy=30)

        self.history = {'grid': [self.grid.copy()], 'acid': [self.acid_grid.copy()], 'yeast': [yeast_count], 'lactic': [lactic_count]}

    def _spawn(self, etype, grid_val, init_energy):
        while True:
            x = random.randint(0, self.width - 1)
            y = random.randint(0, self.height - 1)
            if self.grid[y, x] == 0:

                self.entities[len(self.entities)] = {'type': etype, 'x': x, 'y': y, 'energy': init_energy, 'age': 0}
                self.grid[y, x] = grid_val
                break

    def _get_neighbors(self, x, y, radius=1):
        neighbors = []
        for dx in range(-radius, radius + 1):
            for dy in range(-radius, radius + 1):
                if dx == 0 and dy == 0: continue
                nx, ny = (x + dx) % self.width, (y + dy) % self.height
                neighbors.append((nx, ny))
        return neighbors

    def _get_empty_healthy(self, x, y):
        return [(nx, ny) for nx, ny in self._get_neighbors(x, y, 1) if self.grid[ny, nx] == 0 and self.health_grid[ny, nx] > 0]



    def step(self):
        new_ents = {}
        to_remove = set()
        self.acid_grid = self.acid_grid * 0.93
        self.acid_grid[self.acid_grid < 0.1] = 0.0

        for eid, ent in self.entities.items():
            if eid in to_remove: continue
            x, y = ent['x'], ent['y']

            if self.health_grid[y, x] <= 0 and ent['type'] == 'yeast':
                to_remove.add(eid)
                self.grid[y, x] = 0
                continue

            ent['age'] += 1






            if ent['type'] == 'lactic':
                if ent['age'] < 20:
                    lactic_death_chance = 0.01
                elif ent['age'] < 40:
                    lactic_death_chance = 0.04
                else:
                    lactic_death_chance = 0.10


                if self.health_grid[y, x] < 40:
                    lactic_death_chance += 0.08


                if random.random() < lactic_death_chance:
                    to_remove.add(eid)
                    self.grid[y, x] = 0
                    continue

            empty = self._get_empty_healthy(x, y)


            if ent['type'] == 'yeast':
                self.health_grid[y, x] -= (6.5 * self.pca_pressure)
                bonus = 1.5 if self.health_grid[y, x] < 50 else 1.0
                ent['energy'] += 1.5 * bonus

                if self.acid_grid[y, x] > 3:
                    ent['energy'] -= 6.0

                if empty:
                    self.grid[y, x] = 0
                    nx, ny = random.choice(empty)
                    ent['x'], ent['y'] = nx, ny
                    self.grid[ny, nx] = 1
                    ent['energy'] -= 0.5

                    if ent['energy'] >= 22:
                        ent['energy'] //= 2
                        spawn = self._get_empty_healthy(nx, ny)
                        if spawn:
                            cx, cy = random.choice(spawn)
                            new_ents[max(list(self.entities.keys()) + list(new_ents.keys()) + [0]) + 1] = {'type': 'yeast', 'x': cx, 'y': cy, 'energy': ent['energy'], 'age': 0}
                            self.grid[cy, cx] = 1
                if ent['age'] >= 60:
                    if random.random() < 0.10:
                        to_remove.add(eid)
                        self.grid[y, x] = 0
                        continue

            elif ent['type'] == 'lactic':
                for nx, ny in self._get_neighbors(x, y, 1):
                    self.acid_grid[ny, nx] += 1.0

                decay_rate = 0.8
                if self.health_grid[y, x] > 50:
                    decay_rate -= 0.3

                ent['energy'] -= decay_rate

                if empty and ent['energy'] > 0:
                    self.grid[y, x] = 0
                    nx, ny = random.choice(empty)
                    ent['x'], ent['y'] = nx, ny
                    self.grid[ny, nx] = 2

            if ent['energy'] <= 0:
                to_remove.add(eid)
                self.grid[ent['y'], ent['x']] = 0

        for eid in to_remove:
            if eid in self.entities: del self.entities[eid]
        self.entities.update(new_ents)

        self.grid[self.health_grid <= 0] = 3

        self.history['grid'].append(self.grid.copy())
        self.history['yeast'].append(sum(1 for e in self.entities.values() if e['type'] == 'yeast'))
        self.history['lactic'].append(sum(1 for e in self.entities.values() if e['type'] == 'lactic'))
        self.history['acid'].append(self.acid_grid.copy())

global_pop_history = {'yeast': [], 'lactic': []}
def run_and_save_spatial(yeast_init, lactic_init, steps):
    global global_pop_history
    emp_pressure = calculate_empirical_pca_pressure()
    model = MicrobialCompetitionABM(width=40, height=40, yeast_count=int(yeast_init), lactic_count=int(lactic_init), pca_pressure=emp_pressure)

    output_video_path = "./spatial_simulation.mp4"
    frames = []
    cmap = ListedColormap(['#f8f8f8', '#ef4444', '#3b82f6', '#1f2937'])

    for current_step in range(int(steps)):
        model.step()

        fig, ax = plt.subplots(figsize=(6, 6))
        ax.imshow(model.grid, cmap=cmap, vmin=0, vmax=3)

        if np.max(model.acid_grid) > 0:
            ax.imshow(model.acid_grid, cmap='YlOrBr', alpha=0.3, vmin=0, vmax=10)

        ax.axis('off')

        fig.canvas.draw()
        rgba_buffer = fig.canvas.buffer_rgba()
        rgb_image = np.asarray(rgba_buffer)[:, :, :3]

        frames.append(rgb_image)
        plt.close(fig)

    imageio.mimsave(output_video_path, frames, fps=5, macro_block_size=None)

    global_pop_history['yeast'] = model.history['yeast']
    global_pop_history['lactic'] = model.history['lactic']

    df_pop = pd.DataFrame({
        'Step': range(len(global_pop_history['yeast'])),
        'Total Bacteria': global_pop_history['yeast'],
        'Lactic': global_pop_history['lactic']
    })
    fig_line = px.line(df_pop, x='Step', y=['Total Bacteria', 'Lactic'],
                       color_discrete_map={'Total Bacteria': '#ef4444', 'Lactic': '#3b82f6'},
                       title="Population Dynamics")
    fig_line.update_layout(
        xaxis=dict(range=[0, int(steps)]),
        yaxis_title="Population Size",
        legend_title_text="Microbe Type"
    )
    return output_video_path, fig_line
```

```python
def get_dashboard_metadata():
    final_folder = "./Data/Data"
    if not os.path.exists(final_folder): return "N/A", 0
    excel_files = [f for f in os.listdir(final_folder) if f.endswith(".xlsx")]
    if not excel_files: return "N/A", 0
    file_path = os.path.join(final_folder, excel_files[0])
    mtime = os.path.getmtime(file_path)
    return datetime.datetime.fromtimestamp(mtime).strftime('%Y-%m-%d %H:%M'), len(pd.ExcelFile(file_path).sheet_names)

def get_fig_from_map(selected_metric, data_map):
    data_list = []
    for (day, treatment), obj in data_map.items():
        val = get_avg_val(getattr(obj, selected_metric, None))
        if val is not None: data_list.append({"day": day, "treatment": treatment, "value": val})

    df = pd.DataFrame(data_list)
    if df.empty: return px.line(title="No Data")

    df = df.sort_values(by="day")
    df['Data Type'] = df['day'].apply(lambda d: 'Real Data' if d <= MAX_REAL_DAY else 'Prediction')
    fig = px.line(df, x="day", y="value", color="treatment", line_dash="Data Type", markers=True, line_dash_map={'Real Data': 'solid', 'Prediction': 'dash'})
    fig.update_layout(height=600)
    return fig

def update_compare_graph(metric, t1, t2):

    mapping = {"Control": "control", "Empty Paste": "CMC-3-methyl", "Paste A": "CMC-3-methyl+A", "Paste B": "CMC-3-methyl+B", "Paste C": "CMC-3-methyl+C"}
    target_1, target_2 = mapping.get(t1, t1), mapping.get(t2, t2)
    data_list = []
    for (day, treatment), obj in current_hash_data.items():
        if treatment in [target_1, target_2]:
            val = get_avg_val(getattr(obj, metric, None))
            if val is not None: data_list.append({"day": day, "treatment": t1 if treatment == target_1 else t2, "value": val})

    df = pd.DataFrame(data_list)
    if df.empty: return px.line(title="No Data")

    df = df.sort_values(by="day")
    df['Data Type'] = df['day'].apply(lambda d: 'Real Data' if d <= MAX_REAL_DAY else 'Prediction')
    fig = px.line(df, x="day", y="value", color="treatment", line_dash="Data Type", markers=True, title=f"Compare: {t1} against {t2}", line_dash_map={'Real Data': 'solid', 'Prediction': 'dash'})
    fig.update_layout(height=600, xaxis_title="Time(days)", yaxis_title="value(average)", legend_title_text="Treatment Type")
    return fig


# run_lotka_volterra

def run_lotka_volterra(alpha, beta, gamma, initial_yeast, initial_lactic, days):

    # calculate lotka vultera
    # X =  (Total Bacteria) - pray -the  bacteria that causes dacaying(bad) with its released enzimes or what ever
    # Y =  (Lactic) - predator- the good bacteria that releases increasing ph materials which kills prey

    dt = 0.05
    steps = int(days / dt)

    t_values = np.linspace(0, days, steps)
    yeasts = np.zeros(steps)
    lactics = np.zeros(steps)
    decay = np.zeros(steps)
    decay[0] = 0
    yeasts[0] = initial_yeast
    lactics[0] = initial_lactic
    sensitivity_factor = 0.0001

    for i in range(1, steps):
        x = yeasts[i-1]
        y = lactics[i-1]


        # קצב הגידול של החיידקים מופחת על ידי הלקטיים
        # (1 - beta*y) מייצג את העיכוב שהלקטיים יוצרים
        available_tissue = max(0.0, 1.0 - decay[i-1])
        dx = (alpha * x * (1 - beta * y) * available_tissue) * dt
        #dx = (alpha * x * (1 - beta * y)) * dt

        # קצב הגידול של הלקטיים מוגבל על ידי החיידקים (תחרות)
        current_decay_rate = gamma * (1 + (i / steps) * 1.5)
        #dy = (delta * y * (1 - (y/K))) * dt

        # צבירת הנזק (הריקבון)


        # decay[i] = decay[i-1] + (x * dt * sensitivity_factor)

        if decay[i-1] > 0.1:
            current_decay_rate += 0.2

        dy = (-current_decay_rate * y) * dt


        decay[i] = decay[i-1] + (x * dt * sensitivity_factor)
        #delta  is how well the lactic bacteria is in multiplying(increase the killing rate of yeast)
        #dy is the change in the number of lactic bacteria
        ####
        #that is where the error in the model enters delta * x * y the x here represents that the lactic bacteria eats Total bacteria to grow, but that isnt the case
        #we need to switch to Lotka-Volterra Competition Model competition over same feeding ground

        yeasts[i] = max(0, x + dx)
        lactics[i] = max(0, y + dy)

    # df_sim = pd.DataFrame({
    #     'Time(days)': np.concatenate([t_values, t_values]),
    #     'Colony Size': np.concatenate([yeasts, lactics]),
    #     'Type': ['(Yeasts)'] * steps + ['(Lactic)'] * steps
    # })

    # fig = px.line(df_sim, x='Time(days)', y='Colony Size', color='Type',
    #               title="Symulation of lotka volterra ",
    #               color_discrete_map={'Yeasts': '#eab308', 'Lactic': '#ef4444'})
    # fig.update_layout(height=600)

    df_sim = pd.DataFrame({
        'Time(days)': np.concatenate([t_values, t_values, t_values]),
        'Colony Size': np.concatenate([yeasts, lactics, decay * 50]),
        'Type': ['(Total Bacteria)'] * steps + ['(Lactic)'] * steps + ['(Decay)'] * steps
    })

    fig = px.line(df_sim, x='Time(days)', y='Colony Size', color='Type',
                  title="Lotka-Volterra: Competition & Decay Accumulation",
                  color_discrete_map={'(Total Bacteria)': '#ef4444', '(Lactic)': '#3b82f6', '(Decay)': '#9333ea'})

    fig.update_layout(
        height=600,
        yaxis=dict(range=[0, 1000]),
        xaxis_title="Time (Days)",
        yaxis_title="Colony Size / Decay Scale"
    )
    return fig



gr.close_all()

last_update, num_tabs = get_dashboard_metadata()

ALL_METRICS = [
    ("Weight loss", "weight_loss"),
    ("O2", "o2"),
    ("CO2", "co2"),
    ("TSS", "TSS"),
    ("Texture firmness", "texture_firmness"),
    ("Color", "color"),
    ("Morphology rate", "morphology_rate"),
    ("Bacteria total", "bacteria_total"),

    ("Lactic", "lactic")
]
custom_css = """
.gradio-container { font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; }
#search-btn { background: linear-gradient(to right, #059669, #10b981) !important; border: none !important; box-shadow: 0 4px 6px -1px rgba(0, 0, 0, 0.1); transition: all 0.2s ease-in-out; color: white !important;}
#search-btn:hover { transform: translateY(-2px); box-shadow: 0 10px 15px -3px rgba(0, 0, 0, 0.1); }


#banner-btn { padding: 25px 20px !important; background: linear-gradient(135deg, #064e3b 0%, #059669 100%) !important; border-radius: 16px !important; color: white !important; text-align: center !important; box-shadow: 0 10px 25px -5px rgba(5, 150, 105, 0.4) !important; margin-bottom: 20px !important; border: none !important; font-size: 2.5em !important; font-weight: 800 !important; letter-spacing: -0.5px !important; text-shadow: 0 2px 4px rgba(0,0,0,0.2) !important; width: 100% !important; cursor: pointer !important; transition: all 0.2s ease-in-out !important; }
#banner-btn:hover { transform: scale(1.01) !important; box-shadow: 0 15px 30px -5px rgba(5, 150, 105, 0.5) !important; }
#banner-btn::after { content: 'Dashboard'; display: block; font-size: 0.45em; font-weight: 300; opacity: 0.9; margin-top: 8px; letter-spacing: normal; text-shadow: none; }
"""

theme = gr.themes.Soft(primary_hue="emerald", neutral_hue="slate")

print("Initializing Systems...")

_ = px.line(pd.DataFrame({"x": [1], "y": [1]}), x="x", y="y")

with gr.Blocks(theme=theme, css=custom_css, title="CropSurvive Full Engine") as demo:

    btn_home = gr.Button("CropSurvive", elem_id="banner-btn")

    with gr.Row():

        with gr.Column(scale=1):

            btn_fresh = gr.Button(" Show Freshness")
            btn_heatmap = gr.Button("Decay Heatmap")
            btn_compare = gr.Button("AutoCompare Data")
            btn_predict = gr.Button("Predictive Estimation")
            btn_lv = gr.Button("Lotka Volterra COMPETATIVE ")
            btn_spatial = gr.Button("ABM Simulation")



        with gr.Column(scale=4):

            with gr.Column(visible=True) as page_home:

                header_dynamic = gr.Markdown("# Metric: Oxygen (O2) Over Time")

                with gr.Row():
                    gr.Textbox(value=last_update, label="Last Sync", interactive=False)
                    metric_drp = gr.Dropdown(choices=ALL_METRICS, label="Select Metric", value="o2")
                plot_home = gr.Plot(show_label=False)


            with gr.Column(visible=False) as page_fresh:
                gr.Markdown("# Freshness Score Over Time")

                plot_fresh_line = gr.Plot(show_label=False)

                gr.HTML("""
                    <div style='background-color: var(--background-fill-secondary); padding: 15px; border-radius: 8px; border-left: 4px solid #3b82f6; margin-top: 20px; margin-bottom: 20px;'>
                        <h4 style='margin-top: 0; color: #2563eb;'> Understanding the Relative Freshness Index</h4>
                        <p style='font-size: 14px; margin-bottom: 10px;'>
                            This graph displays the <b>net difference</b> in quality between each treatment and the untreated control group (Score_{Treatment} - Score_{Control}).
                        </p>
                         <p style='font-size: 14px; margin-bottom: 10px;'>
                            All 7 underlying parameters (Weight loss, CO2, O2, Texture, Color, Morphology, Total Bacteria) contribute equally to this index.
                        </p>
                        <ul style='font-size: 14px; margin-top: 0; margin-bottom: 10px; padding-left: 20px;'>
                            <li style='margin-bottom: 5px;'><b>The Control Baseline (0.0):</b> Represented as the zero-line. Everything is evaluated relative to this baseline.</li>
                            <li style='margin-bottom: 5px;'><b>Positive Values (> 0):</b> Indicate success. A positive score proves that the edible coating/paste is actively slowing down decay, keeping the produce fresher while the untreated control degrades at a much faster rate.</li>
                            <li style='margin-bottom: 5px;'><b>Negative Values (< 0):</b> Indicate that the treatment is underperforming compared to the untreated control.</li>
                        </ul>

                        <h4 style='margin-top: 15px; color: #2563eb;'> Mathematical Breakdown (How the Base Score is Built):</h4>
                        <p style='font-size: 14px; margin-bottom: 10px;'>
                            To ensure the score is transparent and scientifically sound, the base freshness score for each specific sample starts at <b>100.0</b> points and is calculated dynamically:
                        </p>
                        <ul style='font-size: 14px; margin-top: 0; margin-bottom: 10px; padding-left: 20px;'>
                            <li style='margin-bottom: 4px;'><b>Equal Weighted Impact:</b> The 100 points are divided equally among the 7 active physiological and microbial metrics. This means each metric has a strict maximum penalty cap of <b>14.3 points</b> (100 / 7).</li>
                            <li style='margin-bottom: 4px;'><b>Relative Day-0 Degradation:</b> For every metric, the engine calculates the pure percentage of change/deterioration relative to its own Day-0 baseline value.</li>
                            <li style='margin-bottom: 4px;'><b>Penalty Capping:</b> If a single parameter experiences extreme degradation (e.g., bacteria count spiking by 400%), its penalty is safely capped at its 14.3-point limit, preventing a single anomaly from crashing the entire index.</li>
                        </ul>


                    </div>
                """)


            with gr.Column(visible=False) as page_heatmap:
                gr.Markdown("# Decay Rate (Heatmap)")

                plot_fresh_heatmap = gr.Plot(show_label=False)
                gr.HTML("""
                <div style='background-color: var(--background-fill-secondary); padding: 15px; border-radius: 8px; border-left: 4px solid #10b981; margin-top: 20px; margin-bottom: 20px;'>
                    <h4 style='margin-top: 0; color: #059669;'> How to Read the Relative Decay Heatmap</h4>
                    <p style='font-size: 14px; margin-bottom: 10px;'>
                        This matrix maps the preservation efficacy over time across all treatments, normalized entirely against the untreated control group.
                    </p>
                    <ul style='font-size: 14px; margin-top: 0; margin-bottom: 10px; padding-left: 20px;'>
                        <li style='margin-bottom: 5px;'><b>Top Row (control = 0.0):</b> The control row serves as the mathematical anchor. Since it is compared to itself, its values remain constantly at 0.0.</li>
                        <li style='margin-bottom: 5px;'><b>Deep Green Tiles (Positive Deltas):</b> Visual proof of shelf-life extension. Shows exactly how many index points the formulation outperforms the control on that specific day by delaying degradation.</li>
                        <li style='margin-bottom: 5px;'><b>Deep Red Tiles (Negative Deltas):</b> Regions where the formulation accelerated tissue decay faster than the natural degradation of the control.</li>
                    </ul>

                    <h4 style='margin-top: 15px; color: #059669;'> Methodological Calculation Framework:</h4>
                    <p style='font-size: 14px; margin-bottom: 10px;'>
                        The data inside these tiles is generated via a two-tier mathematical pipeline:
                    </p>
                    <ol style='font-size: 14px; margin-top: 0; margin-bottom: 10px; padding-left: 20px;'>
                        <li style='margin-bottom: 4px;'><b>Deduction Phase:</b> Each treatment evaluates its current state against its Day-0 data, deducting up to 14.3 points per parameter from a perfect 100.0 score based on its degradation rate.</li>
                        <li style='margin-bottom: 4px;'><b>Normalization Phase:</b> The final value displayed is the absolute treatment score minus the control score for that exact day, translating raw metrics into a 'Preservation Benefit' value.</li>
                    </ol>


                </div>
                """)
            with gr.Column(visible=False) as page_compare:
                gr.Markdown("# AutoCompare Data")
                with gr.Row():
                    compare_metric = gr.Dropdown(choices=ALL_METRICS, label="Choice:", value="o2")
                    compare_t1 = gr.Dropdown(choices=["Control", "Empty Paste", "Paste A", "Paste B", "Paste C"], label="treatment", value="Control")
                    compare_t2 = gr.Dropdown(choices=["Control", "Empty Paste", "Paste A", "Paste B", "Paste C"], label="treatment", value="Paste A")
                compare_plot = gr.Plot(show_label=False)

            with gr.Column(visible=False) as page_predict:
                gr.Markdown("# Predictive Estimation")
                with gr.Row():
                      with gr.Column():

                        target_day_slider = gr.Slider(minimum=MAX_REAL_DAY, maximum=MAX_REAL_DAY + 30, step=1, value=MAX_REAL_DAY + 7, label="Estimate until(day):")
                      with gr.Column():
                        run_prediction_btn = gr.Button("Generate Projection", variant="primary")
                        delete_prediction_btn = gr.Button("Delete Prediction", variant="stop")

                prediction_status = gr.Markdown("")
                preview_plot = gr.Plot(show_label=False)


            with gr.Column(visible=False) as page_lv:
                gr.Markdown("")
                gr.Markdown("""
                <div style='background-color: var(--background-fill-secondary); padding: 14px; border-radius: 12px; border-left: 4px solid #3b82f6; margin-bottom: 15px;'>
                    <h3 style='margin-top: 0; color: #3b82f6; font-size: 1.15em; margin-bottom: 10px;'>Graph Legend</h3>
                    <div style='display: flex; gap: 15px; flex-wrap: wrap; font-size: 13.5px;'>
                        <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 14px; height: 14px; background-color: #ef4444; border-radius: 4px;'></div> <b>Total Bacteria</b> (Population Size)</span>
                        <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 14px; height: 14px; background-color: #3b82f6; border-radius: 4px;'></div> <b>Lactic Protective Bacteria</b> (Population Size)</span>
                        <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 14px; height: 14px; background-color: #9333ea; border-radius: 4px;'></div> <b>Accumulated Decay</b> </span>
                    </div>
                </div>
                """)
                with gr.Row():
                    with gr.Column():
                        lv_alpha = gr.Slider(0.01, 1.0, value=0.1, label="Total Bacteria Growth Rate (Alpha)")
                        lv_beta = gr.Slider(0.001, 0.1, value=0.02, label="Total Bacteria Sensitivity to Acid (Beta)")
                        #lv_delta = gr.Slider(0.001, 0.1, value=0.01, label="Lactic Growth Rate (Delta)")
                        lv_gamma = gr.Slider(0.01, 0.2, value=0.1, label="Lactic Natural Mortality (Gamma)")

                    with gr.Column():
                        lv_yeast_init = gr.Slider(1, 300, value=40, label="Initial Total Bacteria Population")
                        lv_lactic_init = gr.Slider(1, 200, value=100, label="Initial Lactic Population")
                        lv_days = gr.Slider(5, 100, value=50, label="Number of Simulation Days")


                lv_plot = gr.Plot(show_label=False)
                gr.HTML("""
                <div style='background-color: var(--background-fill-secondary); padding: 18px; border-radius: 12px; border-left: 4px solid #64748b; margin-top: 15px;'>
                    <h4 style='margin-top: 0; color: var(--text-color-primary); font-size: 1.1em;'>Underlying Mathematical Model Dynamics (System ODEs):</h4>
                    <ul style='font-size: 13.5px; margin-top: 5px; margin-bottom: 0; padding-left: 20px; line-height: 1.5;'>
                        <li><b>Total Bacteria Growth Inhibition (dx/dt):</b> Total Bacterias multiply exponentially based on the growth rate (Alpha) but are directly suppressed by the protective layer density via acid sensitivity (Beta). Growth is strictly constrained by the remaining available healthy substrate (1 - D).</li>
                        <li><b>Lactic Protective Decay (dy/dt):</b> Lactic bacteria do not replicate on the produce. They experience a baseline natural mortality (Gamma) that accelerates linearly over time due to aging inside the storage package. Mortality further increases when severe decay (D > 0.1) destroys their host polymer matrix.</li>
                        <li><b>Decay Accumulation (dD/dt):</b> Surface tissue rot accumulates as a direct function of Total Bacteria population activity over time. The structural decay curve smoothly levels out as it asymptotically approaches 100% total spoilage (1 - D).</li>
                    </ul>
                <div style='background-color: rgba(37, 99, 235, 0.1); padding: 12px; border-radius: 8px; margin-top: 15px; border: 1px dashed #2563eb;'>
                        <b>Reference: Peleg, M. (2006). Advanced Quantitative Microbiology for Foods and Biosystems: Models for Predicting Growth and Inactivation. CRC Press.</b>
                    </div>
                </div>
                """)
            def set_scenario_base():

                return 20, 60, 150

            def set_scenario_stress():

                return 150, 10, 150

            def set_scenario_recovery():

                return 5, 120, 150
            with gr.Column(visible=False) as page_spatial:
                gr.Markdown("# Agent-Based Modeling (Microbial Competition)")
                gr.Markdown("""
                  <div style='background-color: var(--background-fill-secondary); padding: 18px; border-radius: 12px; border-left: 4px solid #10b981; margin-bottom: 20px;'>
                      <h3 style='margin-top: 0; color: #10b981; font-size: 1.3em;'>Simulation Legend</h3>
                      <div style='display: flex; gap: 15px; flex-wrap: wrap; margin-bottom: 15px; font-size: 14px;'>
                          <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 16px; height: 16px; background-color: #f8f8f8; border: 1px solid #ddd; border-radius: 4px;'></div> <b>Healthy Lettuce Tissue</b></span>
                          <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 16px; height: 16px; background-color: #3b82f6; border-radius: 4px;'></div> <b>Lactic Protective Layer</b></span>
                          <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 16px; height: 16px; background-color: #ef4444; border-radius: 4px;'></div> <b>Total Bacteria</b></span>
                          <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 16px; height: 16px; background-color: #1f2937; border-radius: 4px;'></div> <b>Rotten / Decayed Matrix</b></span>
                          <span style='display: inline-flex; align-items: center; gap: 6px;'><div style='width: 24px; height: 16px; background: linear-gradient(to right, #fff, #fe9929); border: 1px solid #ccc; border-radius: 4px; opacity: 0.7;'></div> <b>Organic Acid Cloud (pH Effect)</b></span>
                      </div>
                  </div>
                """)
                with gr.Row():
                    with gr.Column(scale=1):
                        spatial_yeast = gr.Slider(1, 200, value=20, step=1, label="Initial Total Bacteria")
                        spatial_lactic = gr.Slider(1, 200, value=60, step=1, label="Initial Lactic Bacteria")
                        spatial_steps = gr.Slider(50, 500, value=150, step=10, label="Simulation Steps")
                        with gr.Row():
                            btn_base = gr.Button("Base Scenario", variant="secondary")
                            btn_stress = gr.Button("Stress Scenario", variant="stop")
                            btn_recovery = gr.Button("Recovery Scenario", variant="primary")
                        with gr.Row():
                            run_spatial_btn = gr.Button("Run Custom Simulation", variant="primary",elem_id="search-btn")

                with gr.Row():
                        spatial_grid_plot = gr.Video(autoplay=True, loop=True, show_label=False)
                        spatial_line_plot = gr.Plot(show_label=False)


                gr.Markdown("""
                <div style='background-color: var(--background-fill-secondary); padding: 18px; border-radius: 12px; border-left: 4px solid #64748b; margin-top: 15px;'>
                    <h4 style='margin-top: 0; color: var(--text-color-primary); font-size: 1.1em;'>Ecological Model Rules (Agent Logic):</h4>
                    <ul style='font-size: 13.5px; margin-top: 5px; margin-bottom: 0; padding-left: 20px; line-height: 1.5;'>
                        <li><b>Protective Dynamics:</b> Lactic acid bacteria (Blue) secrete organic acid into their local neighborhood, actively suppressing and draining the metabolic energy of nearby Total bacteria. Lactic agents do not replicate and decay probabilistically based on cellular age and local environment quality.</li>
                        <li><b>Contamination Propagation:</b> Spoilage Total bacteria (Red) consume the vital energy of host lettuce cells at each step (reducing tissue health by 6.5 points). Upon gaining sufficient metabolic energy, they colonize empty healthy adjacent cells (Cellular Automata style).</li>
                        <li><b>Resource Depletion & Collapse:</b> Lettuce cells whose structural health reaches 0 turn permanently into a decayed matrix (Black). Total bacteria trapped in fully degraded clusters without access to raw, healthy substrate will die of starvation, driving a natural ecosystem population crash.</li>
                    </ul>
                </div>
                """)






    # Lotka-Volterra Logic

    lv_inputs = [lv_alpha, lv_beta, lv_gamma, lv_yeast_init, lv_lactic_init, lv_days]
    for slider in lv_inputs:
        slider.change(fn=run_lotka_volterra, inputs=lv_inputs, outputs=lv_plot)
    demo.load(fn=run_lotka_volterra, inputs=lv_inputs, outputs=lv_plot)
    # Compare Logic
    compare_inputs = [compare_metric, compare_t1, compare_t2]
    compare_metric.change(fn=update_compare_graph, inputs=compare_inputs, outputs=compare_plot)
    compare_t1.change(fn=update_compare_graph, inputs=compare_inputs, outputs=compare_plot)
    compare_t2.change(fn=update_compare_graph, inputs=compare_inputs, outputs=compare_plot)
    demo.load(fn=update_compare_graph, inputs=compare_inputs, outputs=compare_plot)

    pages = [page_home, page_fresh, page_heatmap, page_compare, page_predict, page_lv, page_spatial]

    def navigate(page_index):
        return [gr.update(visible=(i == page_index)) for i in range(len(pages))]

    btn_home.click(fn=lambda: navigate(0), inputs=None, outputs=pages).then(
        fn=lambda m: get_fig_from_map(m, current_hash_data), inputs=metric_drp, outputs=plot_home
    )
    btn_fresh.click(fn=lambda: navigate(1), inputs=None, outputs=pages).then(
        fn=update_freshness_graph, inputs=None, outputs=plot_fresh_line
    )
    btn_heatmap.click(fn=lambda: navigate(2), inputs=None, outputs=pages).then(
        fn=generate_freshness_heatmap, inputs=None, outputs=plot_fresh_heatmap
    )
    btn_compare.click(fn=lambda: navigate(3), inputs=None, outputs=pages).then(
        fn=update_compare_graph, inputs=compare_inputs, outputs=compare_plot
    )
    btn_predict.click(fn=lambda: navigate(4), inputs=None, outputs=pages)
    btn_lv.click(fn=lambda: navigate(5), inputs=None, outputs=pages).then(
        fn=run_lotka_volterra, inputs=lv_inputs, outputs=lv_plot
    )
    btn_spatial.click(fn=lambda: navigate(6), inputs=None, outputs=pages)


    # Dashboard Logic
    def update_dashboard(m):
        display_names = {"weight_loss": "Weight Loss", "o2": "Oxygen (O2)", "co2": "Carbon Dioxide (CO2)", "TSS": "TSS", "texture_firmness": "Texture Firmness", "color": "Color", "morphology_rate": "Morphology Rate", "bacteria_total": "Bacteria Total", "lactic": "Lactic"}
        new_title = f"# Metric: {display_names.get(m, m.capitalize())} Over Time"
        return get_fig_from_map(m, current_hash_data), new_title

    metric_drp.change(fn=update_dashboard, inputs=metric_drp, outputs=[plot_home, header_dynamic])


    demo.load(fn=lambda: (get_fig_from_map("o2", current_hash_data), "# Metric: Oxygen (O2) Over Time"),
              inputs=None, outputs=[plot_home, header_dynamic])


    # Predict & Freshness Logic
    run_prediction_btn.click(fn=apply_predictions_from_ui, inputs=[target_day_slider], outputs=[prediction_status, preview_plot])
    delete_prediction_btn.click(fn=reset_predictions_from_ui, inputs=None, outputs=[prediction_status, preview_plot])
    demo.load(fn=lambda: (update_freshness_graph(), generate_freshness_heatmap(), get_fig_from_map("o2", current_hash_data)), inputs=None, outputs=[plot_fresh_line, plot_fresh_heatmap, preview_plot])

    spatial_inputs = [spatial_yeast, spatial_lactic, spatial_steps]
    demo.load(fn=run_and_save_spatial, inputs=spatial_inputs, outputs=[spatial_grid_plot, spatial_line_plot])

    btn_base.click(
        fn=set_scenario_base,
        outputs=[spatial_yeast, spatial_lactic, spatial_steps]
    ).then(
        fn=run_and_save_spatial,
        inputs=[spatial_yeast, spatial_lactic, spatial_steps],
        outputs=[spatial_grid_plot, spatial_line_plot]
    )


    btn_stress.click(
        fn=set_scenario_stress,
        outputs=[spatial_yeast, spatial_lactic, spatial_steps]
    ).then(
        fn=run_and_save_spatial,
        inputs=[spatial_yeast, spatial_lactic, spatial_steps],
        outputs=[spatial_grid_plot, spatial_line_plot]
    )


    btn_recovery.click(
        fn=set_scenario_recovery,
        outputs=[spatial_yeast, spatial_lactic, spatial_steps]
    ).then(
        fn=run_and_save_spatial,
        inputs=[spatial_yeast, spatial_lactic, spatial_steps],
        outputs=[spatial_grid_plot, spatial_line_plot]
    )
    run_spatial_btn.click(
        fn=run_and_save_spatial,
        inputs=[spatial_yeast, spatial_lactic, spatial_steps],
        outputs=[spatial_grid_plot, spatial_line_plot]
    )




demo.launch(share=True, debug=True)
```
